# Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgba
from matplotlib.lines import Line2D
from pathlib import Path
```

```python
input_csv_name = "combined_segmentation_quality_v2.csv"
df = pd.read_csv(input_csv_name)
print(df.shape)
df.tail(2)
```

# Metrics by dataset

```python
output_file_name = "metrics_by_dataset.pdf"

metrics_list = [
    "precision", "recall", "f1", "ap50", "ap75", "ap90", "count_error",
    "b_prec_50", "b_prec_75", "b_prec_90",
    "b_rec_50", "b_rec_75", "b_rec_90",
    "b_f1_50", "b_f1_75", "b_f1_90",
    "mean_iou", "median_iou", "mean_dice",
    "foreground_iou", "foreground_dice", "aji_custom", "aji_official",
    "n_matched_groups", "n_split_groups", "n_merged_groups",
    "n_complex_groups", "n_missed_groups", "n_artifact_groups",
]

in_train = ["official_dataset/train", "official_dataset/train_cyto2"]
blue = "tab:blue"
orange = "tab:orange"

def short(name):
    parts = name.split("/")
    return "official/" + parts[1] if parts[0] == "official_dataset" else parts[0]

datasets = list(df["dataset"].unique())
colors = [blue if d in in_train else orange for d in datasets]
labels = [f"{short(d)}\n(n={(df['dataset'] == d).sum()})" for d in datasets]

rng = np.random.default_rng(0)
fig, axes = plt.subplots(len(metrics_list), 1, figsize=(8, 3.5 * len(metrics_list)))   # no sharex

for ax, metric in zip(axes, metrics_list):
    for i, (d, c) in enumerate(zip(datasets, colors)):
        y = df.loc[df["dataset"] == d, metric].dropna()
        x = i + rng.uniform(-0.25, 0.25, len(y))                  # jitter
        ax.scatter(x, y, s=20, facecolor=to_rgba(c, 0.3), edgecolor=c, linewidth=1)
        ax.hlines(y.median(), i - 0.35, i + 0.35, color="black", linewidth=2)  # median
    ax.set_ylabel(metric)
    ax.set_xticks(range(len(datasets)))                           # names on every subplot
    ax.set_xticklabels(labels, rotation=30, ha="right")

# Legend on top of the first subplot
handles = [
    Line2D([], [], marker="o", linestyle="", markerfacecolor=to_rgba(blue, 0.3),
           markeredgecolor=blue, label="in train set"),
    Line2D([], [], marker="o", linestyle="", markerfacecolor=to_rgba(orange, 0.3),
           markeredgecolor=orange, label="not in train set"),
    Line2D([], [], color="black", linewidth=2, label="median"),
]
axes[0].legend(handles=handles, loc="lower center", bbox_to_anchor=(0.5, 1.02),
               ncol=3, frameon=False)

fig.tight_layout()
fig.savefig(output_file_name)   # PDF, because of the .pdf extension
plt.show()
```

Conclusions: performance between the datasets is mostly comparable, expect for the flow_chamber_dataset which performs quite worse on some metrics.
